```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_csv("Algerian_forest_fires_dataset.csv")
```

```python
df.info()
```

Data cleaning

```python
#missing values

df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

The dataset is coverted into set based on region from 122th index ,we can make new column basedon the region  
1: "Bejaia Region datset"  
2: "Side-Bel Abbes Region dataset  

Add new column with region

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
```

```python
df.info()
```

```python
df['Region']=df['Region'].astype(int)
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
## remove the null values

df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.loc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.loc[[122]]
```

```python
df.columns
```

```python
## removing the spaces in the column name

df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
df.info()
```

```python
## change the required columns are integer

df[['day','month','year','Temperature','RH','Ws']]=df[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
# changing the other column to float datatype
columns=df[['Rain','FFMC','DMC','DC','ISI','BUI','FWI']]
```

```python
for col in columns:
    df[col]=df[col].astype(float)
```

```python
df.info()
```

```python
## let's save the cleaned dataset

df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```

## EDA: Exploratory Data Analysis

```python
#drop day,month,year becuase its not required
df_copy=df.drop(['day','month','year'],axis=1)
```

```python
df_copy.head()
```

```python
## Encoding of the categories in classes

df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
## This is my category in Classes
df_copy['Classes'].value_counts()
```

```python
## plotdensity plot for all feature

plt.style.use('ggplot')
df_copy.hist(bins=50,figsize=(20,20),color='purple')
plt.show()
```

```python
## Percentage for pie chart

percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
# ploting pie chart

classlabels=["Fire","Not Fire"]
plt.Figure(figsize=(15,20))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title("Pie chart of classes")
plt.show()
```

```python
## correlation

df_copy.corr()
```

```python

sns.heatmap(df_copy.corr(),annot=True)
```

```python
## box plot

sns.boxplot(df_copy['FWI'],color='green')
```

```python
df['Classes']=np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
## monthly fire analysis
dftemp=df.loc[df['Region']==1]
plt.figure(figsize=(13,6))
sns.set_style('darkgrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.xlabel("Month",weight='bold')
plt.ylabel("Numberof fires",weight='bold')
plt.title("Fire Analysis Of Sidi-Bel Region",weight='bold')
plt.show()
```

```python
## monthly fire analysis
dftemp=df.loc[df['Region']==0]
plt.figure(figsize=(13,6))
sns.set_style('darkgrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.xlabel("Month",weight='bold')
plt.ylabel("Numberof fires",weight='bold')
plt.title("Fire Analysis Of Bejaia Region",weight='bold')
plt.show()
```
